# Polystyrene tacticity with CAPS

One chain to look at, then a small sweep over tacticity: cells grown, relaxed and analysed, three seeds each. `caps` is the Python package in `data/python` (set `PYTHONPATH` to it, or run from the Studio's Python).

```python
import caps

doc = caps.polymer("*CC(*)c1ccccc1", dp=40, tacticity="atactic", seed=5)
doc.view(style="ball-and-stick")
```

```python
# a small sweep (about two minutes on a laptop): 3 tacticities × 3 seeds, 8 chains of DP 20, relaxed.
# Density stays the grown 0.5 g/cm³ unless the cells run NPT (npt_ps=…); tg needs a cooling scan (tg={"t_start": 500, "t_end": 250}).
sweep = caps.sweep.run("*CC(*)c1ccccc1", tacticities=("isotactic", "syndiotactic", "atactic"), dps=(20,), seeds=(1, 2, 3), chains=8)

cells = [caps.sweep.result(t) for t in ("isotactic", "syndiotactic", "atactic")]
caps.table(cells, ["density", "tg", "c_inf"])
```
